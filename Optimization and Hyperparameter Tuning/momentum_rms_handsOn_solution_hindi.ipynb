placeholders नाम का एक function define करें, जो दो placeholders return करे — एक input data के लिए (A_0) और एक output data के लिए (Y)।
- Placeholders का datatype **float64** रखें
- Parameters - num_features
- Returns - A_0 की shape (num_feature, None) और Y की shape (1, None) होनी चाहिए


```python
def placeholders(num_features):
    # placeholder एक तरह का 'खाली डब्बा' (empty container) है, जिसमें आगे चलकर training के समय असली data भरा जाएगा
    A_0 = tf.placeholder(dtype = tf.float64, shape = ([num_features,None]))    # A_0 -> input data के लिए placeholder, shape = (features, कितने भी samples)
    Y = tf.placeholder(dtype = tf.float64, shape = ([1,None]))                 # Y -> असली/actual output (target) के लिए placeholder, shape = (1, कितने भी samples)
    return A_0,Y                                                                # दोनों placeholders को वापस (return) कर रहे हैं
```

initialize_parameters_deep() नाम का function define करें, जो हर layer के लिए weights और bias initialize करे
- Weights initialize करने के लिए tf.random_normal_initializer() और bias के लिए tf.zeros() इस्तेमाल करें। datatype **float64** रखें
- Parameters - layer_dims
- Returns - weights और bias की dictionary


```python
def initialize_parameters_deep(layer_dims):
    tf.set_random_seed(1)                             # random seed फिक्स (fix) कर रहे हैं, ताकि हर बार code चलाने पर same (एक जैसे) random values मिलें
    L = len(layer_dims)                                # L = कुल कितनी layers हैं (layer_dims की length)
    parameters = {}                                    # एक खाली dictionary बना रहे हैं, जिसमें सारे W (weight) और b (bias) रखे जाएँगे
    for l in range(1,L):                               # हर layer (1 से L-1 तक) के लिए loop चला रहे हैं
        ###Start code here
        parameters['W' + str(l)] = tf.Variable(tf.random_normal_initializer()([layer_dims[l], layer_dims[l-1]], dtype=tf.float64))
        # ऊपर वाली line: 'W1', 'W2'... नाम से weight matrix बना रहे हैं, random normal values से भरकर, shape = (current layer size, पिछली layer size)
        parameters['b' + str(l)] = tf.Variable(tf.zeros([layer_dims[l], 1], dtype=tf.float64))
        # ऊपर वाली line: 'b1', 'b2'... नाम से bias vector बना रहे हैं, शुरुआत में सारे 0 से भरा हुआ, shape = (current layer size, 1)
        #End code
    return parameters                                  # सारे weights और biases की dictionary वापस (return) कर रहे हैं
```

linear_forward_prop() नाम का function define करें, जो किसी एक दी हुई layer के लिए forward propagation करे।
- Parameters: A_prev (पिछली layer का output), W (current layer की weight matrix), b (current layer का bias vector), activation (current layer के output के लिए किस तरह का activation इस्तेमाल करना है)
- Returns: A (current layer का output)
- Hidden layers के लिए relu activation इस्तेमाल करें, और आखिरी (final) output layer के लिए output बिना activation लगाए वापस करें यानी अगर activation sigmoid है


```python
def linear_forward_prop(A_prev,W,b, activation):
    Z = tf.add(tf.matmul(W, A_prev), b)      # Z = (W * A_prev) + b  -> यह एक layer की linear calculation है (matrix multiplication फिर bias जोड़ना)
    if activation == "sigmoid":
        A = Z                                # अगर activation "sigmoid" है, तो हम Z को बिना activation लगाए ही आगे भेज रहे हैं (क्योंकि cost function में sigmoid पहले से शामिल है)
    elif activation == "relu":
        A = tf.nn.relu(Z)                    # अगर activation "relu" है, तो Z पर ReLU activation function लगा रहे हैं (negative values को 0 कर देता है)
    return A                                 # इस layer का output (A) वापस (return) कर रहे हैं
```

पूरे network के लिए forward propagation को l_layer_forward() के रूप में define करें
- Parameters: A_0 (input data), parameters (weights और bias की dictionary)
- Returns: A (आखिरी layer का output)


```python
def l_layer_forwardProp(A_0, parameters):
    A = A_0                                  # A को input data (A_0) से शुरू कर रहे हैं
    L = len(parameters)//2                   # कुल कितनी layers हैं, यह parameters dictionary के size से पता कर रहे हैं (हर layer के W और b दोनों हैं, इसलिए //2)
    for l in range(1,L):                     # पहली से लेकर (L-1)वीं layer तक loop चला रहे हैं — यह सारी hidden layers हैं
        A_prev = A                           # पिछली layer का output अगली layer के input के रूप में सेव कर रहे हैं
    ###Start code
        A = linear_forward_prop(A_prev, parameters['W' + str(l)], parameters['b' + str(l)], "relu")   #call linear forward prop with relu activation
        # ऊपर वाली line: hidden layer के लिए forward propagation चला रहे हैं, activation = "relu"
    A = linear_forward_prop(A, parameters['W' + str(L)], parameters['b' + str(L)], "sigmoid")          #call linear forward prop with sigmoid activation
    # ऊपर वाली line: सबसे आखिरी (output) layer के लिए forward propagation, activation = "sigmoid" (पर बिना activation लगाए raw output देंगे)
    ###End code
    return A                                 # पूरे network का आखिरी output (Z_final) वापस (return) कर रहे हैं
```

- Cost function define करें
- Parameters:
  - Z_final: आखिरी layer से आया output
  - Y: असली (actual) output
  - parameters: weights और bias की dictionary
  - regularization: boolean (True/False)
  - lambd: regularization parameter
- पहले TensorFlow के sigmoid_cross_entropy function से असली cost define करें
- अगर **regularization == True** है, तो असली cost में regularization term जोड़ें


```python
def final_cost(Z_final, Y , parameters, regularization = False, lambd = 0):
    cost = tf.nn.sigmoid_cross_entropy_with_logits(logits=Z_final,labels=Y)
    # ऊपर वाली line: असली cost (loss) निकाल रहे हैं — यह देखता है कि model का prediction (Z_final) असली answer (Y) से कितना अलग है
    if regularization:                        # अगर regularization चाहिए (True है), तो...
        reg_term = 0                          # regularization term को 0 से शुरू कर रहे हैं
        L = len(parameters)//2                # कुल layers की संख्या निकाल रहे हैं
        for l in range(1,L+1):                # हर layer के लिए loop चला रहे हैं
            ###Start code
            reg_term += tf.nn.l2_loss(parameters['W' + str(l)])                #add L2 loss term
            # ऊपर वाली line: हर layer के weight (W) का L2 loss (squares का sum, आधा) जोड़ रहे हैं — इससे weights बहुत बड़े होने से बचते हैं (overfitting कम होता है)
            ###End code
        cost = cost + (lambd/2) * reg_term    # असली cost में regularization term जोड़ रहे हैं, lambd (regularization strength) के हिसाब से
    return tf.reduce_mean(cost)               # सारे samples के cost का औसत (average/mean) निकालकर वापस (return) कर रहे हैं
```

Mini-batches generate करने के लिए function define करें। **ज़रूरी: random indices generate करने के लिए np.random.permutation इस्तेमाल करें**


```python
import numpy as np
def random_samples_minibatch(X, Y, batch_size, seed = 1):
    np.random.seed(seed)                                      # random seed फिक्स कर रहे हैं ताकि हर बार same shuffle मिले
    ###Start code
    m = X.shape[1]                                            #number of samples
    # ऊपर: कुल कितने samples हैं (X की दूसरी dimension), यह m में रख रहे हैं
    num_batches = int(m // batch_size)                        #number of batches derived from batch_size
    # ऊपर: कुल कितने पूरे (complete) mini-batch बन सकते हैं, यह निकाल रहे हैं (integer division)
    ###End code
    indices = np.random.permutation(m)                    # generate ramdom indicies
    # ऊपर: 0 से m-1 तक के नंबरों को randomly मिला (shuffle कर) रहे हैं, ताकि data का क्रम (order) random हो जाए
    shuffle_X = X[:,indices]                                  # X को इन्हीं random indices के हिसाब से फिर से arrange (shuffle) कर रहे हैं
    shuffle_Y = Y[:,indices]                                  # Y को भी उसी क्रम में shuffle कर रहे हैं (ताकि X और Y आपस में match रहें)
    mini_batches = []                                         # एक खाली list बना रहे हैं, जिसमें सारे mini-batches डालेंगे
    
    #generate minibatch
    for i in range(num_batches):                              # हर पूरे batch के लिए loop चला रहे हैं
        X_batch = shuffle_X[:, i*batch_size : (i+1)*batch_size]   # X का एक हिस्सा (slice) निकाल रहे हैं — यह एक mini-batch का input है
        Y_batch = shuffle_Y[:, i*batch_size : (i+1)*batch_size]   # Y का वही corresponding हिस्सा निकाल रहे हैं — यह उस mini-batch का target है
        
        assert X_batch.shape == (X.shape[0], batch_size)      # जाँच: X_batch की shape सही है या नहीं
        assert Y_batch.shape == (Y.shape[0], batch_size)      # जाँच: Y_batch की shape सही है या नहीं
        
        mini_batches.append((X_batch, Y_batch))               # (X_batch, Y_batch) की जोड़ी को list में जोड़ (add कर) रहे हैं
    
    #generate batch with remaining number of samples
    if m % batch_size != 0:                                   # अगर कुछ samples बच गए हैं (जो पूरे batch_size में नहीं बँटे), तो...
        X_batch = shuffle_X[:, num_batches*batch_size : m]    # बचे हुए (leftover) X samples निकाल रहे हैं
        Y_batch = shuffle_Y[:, num_batches*batch_size : m]    # बचे हुए Y samples निकाल रहे हैं
        mini_batches.append((X_batch, Y_batch))               # इस आखिरी छोटे batch को भी list में जोड़ रहे हैं
    return mini_batches                                       # सारे mini-batches की list वापस (return) कर रहे हैं
```

Network को mini-batch की मदद से train करने के लिए model define करें
- Parameters:
  - X_train, Y_train: input और target data
  - layer_dims: network का configuration
  - learning_rate
  - optimizer
  - num_iter: कितने epochs (iterations) चलाने हैं
  - mini_batch_size: हर mini-batch में कितने samples लेने हैं
- Return: trained parameters की dictionary


```python
def model(X_train,Y_train, layer_dims, learning_rate, optimizer ,num_iter, mini_batch_size):
    tf.reset_default_graph()                          # पुराना TensorFlow graph मिटाकर (clear करके) नया शुरू कर रहे हैं, ताकि पुराने variables टकराएँ (clash) ना करें
    num_features, num_samples = X_train.shape          # X_train की shape से features और samples की संख्या निकाल रहे हैं
    ###Start code
    A_0, Y = placeholders(num_features)                        #call placeholder function to initialize placeholders A_0 and Y
    # ऊपर: input (A_0) और output (Y) के लिए placeholders बना रहे हैं
    parameters = initialize_parameters_deep(layer_dims)         #Initialse Weights and bias using initialize_parameters
    # ऊपर: सारी layers के weights (W) और bias (b) को initialize (शुरुआती value देकर) कर रहे हैं
    Z_final = l_layer_forwardProp(A_0, parameters)               #call the function l_layer_forwardProp() to define the final output
    # ऊपर: पूरे network से data आगे बढ़ाकर (forward propagation करके) आखिरी output निकाल रहे हैं
    
    cost = final_cost(Z_final, Y, parameters, regularization=True)      #call the final_cost function with regularization set TRUE
    # ऊपर: model का cost (loss) निकाल रहे हैं, regularization को True रखकर (overfitting रोकने के लिए)
    ###End code
    
    ###Start code
    if optimizer == "momentum":
        train_net = tf.train.MomentumOptimizer(learning_rate=learning_rate, momentum=0.9).minimize(cost)                  #call tensorflow's momentum optimizer with momentum = 0.9
        # ऊपर: अगर optimizer "momentum" चुना है, तो Momentum Optimizer से cost को कम (minimize) करने वाला operation बना रहे हैं
    elif optimizer == "rmsProp":
        train_net = tf.train.RMSPropOptimizer(learning_rate=learning_rate, decay=0.999).minimize(cost)                    #call tensorflow's RMS optimiser with decay = 0.999
        # ऊपर: अगर optimizer "rmsProp" चुना है, तो RMSProp Optimizer से cost को minimize करने वाला operation बना रहे हैं
    elif optimizer == "adam":
        train_net = tf.train.AdamOptimizer(learning_rate=learning_rate, beta1=0.9, beta2=0.999).minimize(cost)            ##call tensorflow's adam optimizer with beta1 = 0.9, beta2 = 0.999
        # ऊपर: अगर optimizer "adam" चुना है, तो Adam Optimizer से cost को minimize करने वाला operation बना रहे हैं
    ###End code
    seed = 1                                            # mini-batch बनाने के लिए शुरुआती seed value
    num_minibatches = int(num_samples / mini_batch_size)   # कुल कितने mini-batches बनेंगे, यह निकाल रहे हैं
    init = tf.global_variables_initializer()            # TensorFlow के सारे variables (W, b आदि) को initialize करने वाला operation बना रहे हैं
    costs = []                                           # हर epoch का cost store करने के लिए खाली list
    with tf.Session() as sess:                           # TensorFlow Session शुरू कर रहे हैं (यहीं असली calculation चलती है)
        sess.run(init)                                   # सारे variables को उनकी शुरुआती value से set (initialize) कर रहे हैं
        for epoch in range(num_iter):                    # हर epoch (एक बार पूरे data पर training) के लिए loop
            epoch_cost = 0                                # इस epoch का कुल cost 0 से शुरू कर रहे हैं
            ###Start code
            mini_batches = random_samples_minibatch(X_train, Y_train, mini_batch_size, seed)               #call random_sample_minibatch to return minibatches
            # ऊपर: पूरे training data से random mini-batches बना रहे हैं
            ###End code
            seed = seed + 1                               # अगली बार अलग shuffle मिले, इसलिए seed को बदल (increase कर) रहे हैं
            
            #perform gradient descent for each mini-batch
            for mini_batch in mini_batches:               # हर mini-batch के लिए loop चला रहे हैं
                ###Start code
                X_batch, Y_batch = mini_batch             #assign minibatch
                # ऊपर: mini_batch (जोड़ी) से X_batch और Y_batch अलग-अलग निकाल रहे हैं
                ###End code
                _,mini_batch_cost = sess.run([train_net, cost], feed_dict={A_0: X_batch, Y: Y_batch})
                # ऊपर: इस mini-batch पर training चलाकर (train_net) weights update कर रहे हैं, साथ ही इस batch का cost भी निकाल रहे हैं
                epoch_cost += mini_batch_cost/num_minibatches   # इस batch का cost जोड़कर epoch के कुल (average) cost में शामिल कर रहे हैं
            
            if epoch % 2 == 0:                            # हर 2 epoch के बाद...
                costs.append(epoch_cost)                  # उस epoch का cost list में save कर रहे हैं (graph बनाने के लिए)
            if epoch % 10 == 0:                           # हर 10 epoch के बाद...
                print(epoch_cost)                         # cost को print (दिखा) रहे हैं, ताकि training कैसी चल रही है पता चले
        optimization.save_ans7(epoch_cost, optimizer)     # आखिरी epoch का cost checker module में save कर रहे हैं (grading के लिए)
        plt.ylim(0 ,2, 0.0001)                             # graph की y-axis की सीमा (limit) सेट कर रहे हैं
        plt.xlabel("epoches per 2")                        # x-axis का नाम (label) दे रहे हैं
        plt.ylabel("cost")                                 # y-axis का नाम (label) दे रहे हैं
        plt.plot(costs)                                    # सारे saved costs को graph में plot कर रहे हैं
        plt.show()                                         # graph दिखा रहे हैं
        params = sess.run(parameters)                      # trained (train किए हुए) weights और bias को असली numbers के रूप में निकाल रहे हैं
    return params                                          # trained parameters वापस (return) कर रहे हैं
```
