placeholders नाम का function define करें जो दो placeholders return करे - एक input data के लिए A_0 और एक output data के लिए Y
- Placeholders का datatype float64 set करें
- parameters - num_features
- Returns - A_0 जिसका shape (num_feature, None) है और Y जिसका shape (1,None) है

```python
def placeholders(num_features, num_classes):    #function define करें जो num_features और num_classes parameters लेगा
    A_0 = tf.placeholder(dtype = tf.float64, shape = ([num_features,None]))   #input data के लिए placeholder A_0 बनाएं, shape में None का मतलब है samples की संख्या run के समय तय होगी
    Y = tf.placeholder(dtype = tf.float64, shape = ([num_classes,None]))      #target/output data के लिए placeholder Y बनाएं, shape (num_classes, samples) होगा
    return A_0,Y     #दोनों placeholders को return करें
```

initialize_parameters_deep() नाम का function define करें जो हर layer के लिए weights और bias initialize करे
- weights और bias initialize करने के लिए tf.get_variable का उपयोग करें, datatype float64 set करें
- ध्यान रखें weights के लिए xavier initialization का उपयोग करें और bias को zeros से initialize करें
- Parameters - layer_dims
- Returns - weights और bias की dictionary

```python
def initialize_parameters_deep(layer_dims):     #function define करें जो layer_dims (हर layer में nodes की संख्या की list) लेगा
    tf.set_random_seed(1)     #random seed set करें ताकि हर बार same random values initialize हों (reproducibility के लिए)
    L = len(layer_dims)       #layer_dims की length से total layers की संख्या (input सहित) पता करें
    parameters = {}           #सभी weights और bias को store करने के लिए एक empty dictionary बनाएं
    for l in range(1,L):      #layer 1 से लेकर last layer तक loop चलाएं (0th layer input होता है, उसका कोई weight नहीं होता)
        parameters['W' + str(l)] = tf.get_variable("W" + str(l), shape=[layer_dims[l], layer_dims[l-1]], dtype = tf.float64,
                                   #इस layer 'l' के लिए weight matrix W बनाएं, shape (current layer nodes, previous layer nodes) होगा
                                   initializer=tf.contrib.layers.xavier_initializer())   #weights को xavier initialization technique से initialize करें, इससे training stable रहती है
        parameters['b' + str(l)] = tf.get_variable("b"+ str(l), shape = [layer_dims[l], 1], dtype= tf.float64, initializer= tf.zeros_initializer() )   #इस layer के लिए bias vector b बनाएं, shape (current layer nodes, 1), zeros से initialize करें
    return parameters      #सभी layers के weights और bias वाली dictionary return करें
```

Define functon named linear_forward_prop() to define forward propagation for a given layer.
linear_forward_prop() नाम का function define करें जो किसी given layer के लिए forward propagation define करे
- parameters: A_prev (पिछली layer का output), W (current layer का weight matrix), b (current layer का bias vector), activation (current layer के output के लिए किस तरह का activation उपयोग करना है)  
- returns: A (current layer का output)
- Hidden layers के लिए relu activation उपयोग करें, और अंतिम output layer के लिए output को बिना activation लगाए ही return करें यानी अगर activation 'softmax' है

```python
def linear_forward_prop(A_prev,W,b, activation):   #function define करें जो A_prev, W, b, activation parameters लेगा
    #Start code here 
    Z = tf.add(tf.matmul(W, A_prev), b)   #linear combination Z = W.A_prev + b calculate करें (matmul से matrix multiplication, फिर bias add करें)
    if activation == "softmax":    #अगर activation 'softmax' है (यानी यह final/output layer है)
        A = Z    #output layer के लिए Z को बिना किसी activation के ही A में assign करें (softmax बाद में cost function में लगाया जाएगा)
    elif activation == "relu":    #अगर activation 'relu' है (यानी यह hidden layer है)
        A = tf.nn.relu(Z)    #Z पर relu activation function लगाकर A calculate करें, यह negative values को 0 कर देता है
    return A    #current layer का output A return करें
   #End code
```

पूरे network के लिए forward propagation l_layer_forward() के रूप में आपके लिए define किया गया है। यहाँ हम किसी regularization technique का उपयोग नहीं कर रहे हैं।

```python
def l_layer_forwardProp(A_0, parameters):    #function define करें जो input A_0 और सभी parameters (weights, bias) लेगा
    A = A_0    #A को शुरुआत में input data A_0 के बराबर set करें
    L = len(parameters)//2    #parameters dictionary में हर layer के 2 items (W और b) होते हैं, इसलिए 2 से divide करके total layers की संख्या L निकालें
    for l in range(1,L):    #पहली hidden layer से लेकर second-last layer तक loop चलाएं
        A_prev = A    #पिछली layer के output को A_prev में store करें
        A = linear_forward_prop(A_prev, parameters['W' + str(l)], parameters['b' + str(l)], "relu")   #relu activation के साथ current layer का forward propagation करें
    A = linear_forward_prop(A, parameters['W' + str(L)], parameters['b' + str(L)], "softmax" )   #अंतिम (output) layer के लिए softmax (यानी बिना activation लगाए raw Z) के साथ forward propagation करें
    return A    #network का final output A return करें
```

- cost function को final_cost() के रूप में define करें
- softmax_cross_entropy_with_logits() function से cost calculate करने के लिए logits और target value का shape (samples की संख्या, class की संख्या) होना चाहिए      
- चूंकि Z_final और Y tensorflow object हैं, इसलिए softmax_cross_entropy_with_logits() को feed करने से पहले हम tf.transpose() का उपयोग कर रहे हैं

```python
def final_cost(Z_final, Y ):   #function define करें जो network का final output Z_final और actual target Y लेगा
    logits = tf.transpose(Z_final)   #Z_final को transpose करें ताकि shape (classes, samples) से (samples, classes) हो जाए
    labels = tf.transpose(Y)         #Y को भी transpose करें ताकि उसका shape भी (samples, classes) हो जाए
    ###Start code
    cost = tf.nn.softmax_cross_entropy_with_logits(logits=logits, labels=labels)   #tensorflow के softmax_cross_entropy function से हर sample का cost/loss calculate करें
    ###End code
    return tf.reduce_mean(cost)    #सभी samples के cost का average (mean) निकालकर return करें
```

Mini-batches generate करने के लिए function define करें।

```python
import numpy as np    #numpy library import करें - array/numerical operations के लिए, इसे 'np' नाम से इस्तेमाल करेंगे
def random_samples_minibatch(X, Y, batch_size, seed = 1):   #function define करें - X, Y data, batch_size और seed parameters के साथ
    np.random.seed(seed)    #random seed set करें ताकि shuffling reproducible (हर बार same) रहे
    m = X.shape[1]                                 #samples की कुल संख्या निकालें (X के columns की संख्या)
    num_batches = int(m/batch_size)                #batch_size के हिसाब से कुल कितने पूरे (complete) batches बनेंगे, यह निकालें
    indices = np.random.permutation(m)             #0 से m-1 तक के numbers को randomly shuffle करके एक index array बनाएं
    shuffle_X = X[:,indices]    #इन random indices के अनुसार X के columns को shuffle (फेरबदल) करें
    shuffle_Y = Y[:,indices]    #उन्हीं indices से Y के columns को भी उसी क्रम में shuffle करें, ताकि X-Y का mapping सही बना रहे
    mini_batches = []    #सभी mini-batches (X_batch, Y_batch) को store करने के लिए खाली list बनाएं
    
    #generate minibatch    (पूरे size वाले) mini-batches बनाएं
    for i in range(num_batches):    #हर पूरे batch के लिए loop चलाएं
        ##Start code here
        X_batch = shuffle_X[:, i*batch_size : (i+1)*batch_size]    #shuffle किए हुए X से i-वें batch के samples (columns) slice करके निकालें
        Y_batch = shuffle_Y[:, i*batch_size : (i+1)*batch_size]    #shuffle किए हुए Y से भी वही i-वें batch के samples slice करें
        ##End code
        
        assert X_batch.shape == (X.shape[0], batch_size)    #जांचें X_batch का shape (features, batch_size) सही है या नहीं
        assert Y_batch.shape == (Y.shape[0], batch_size)    #जांचें Y_batch का shape (classes, batch_size) सही है या नहीं
        
        mini_batches.append((X_batch, Y_batch))    #इस batch को (X_batch, Y_batch) tuple के रूप में mini_batches list में जोड़ें
    
    #generate batch with remaining number of samples    बचे हुए (remaining) samples से आखिरी छोटा batch बनाएं
    if m % batch_size != 0:    #अगर total samples batch_size से पूरी तरह divide नहीं होते (यानी कुछ samples बच जाते हैं)
        ##Srart code here
        X_batch = shuffle_X[:, num_batches*batch_size :]    #बचे हुए सभी samples को X से slice करके निकालें
        Y_batch = shuffle_Y[:, num_batches*batch_size :]    #बचे हुए सभी samples को Y से भी slice करके निकालें
        ##Srart code here
        mini_batches.append((X_batch, Y_batch))    #इस आखिरी छोटे batch को भी mini_batches list में जोड़ें
    return mini_batches    #सभी mini-batches की list return करें
```

Network को minibatch का उपयोग करके train करने के लिए model define करें
- parameters:
  - X_train, Y_train: input और target data
  - layer_dims: network की configuration
  - learning_rate
  - num_iter: epochs की संख्या
  - mini_batch_size: हर minibatch में कितने samples लिए जाएं इसकी संख्या
- return: trained parameters की dictionary

```python
def model_with_minibatch(X_train,Y_train, layer_dims, learning_rate, num_iter, mini_batch_size):   #model function define करें जो training के सभी parameters लेगा
    
    tf.reset_default_graph()    #tensorflow के default computation graph को reset करें ताकि पुराने variables/graph से टकराव न हो
    num_features, num_samples = X_train.shape    #X_train के shape से features और samples की संख्या निकालें
    num_classes = Y_train.shape[0]    #Y_train के shape से classes की संख्या निकालें
    A_0, Y = placeholders(num_features, num_classes)    #input और output के लिए placeholders बनाएं
    parameters = initialize_parameters_deep(layer_dims)    #सभी layers के weights और bias initialize करें
    Z_final = l_layer_forwardProp(A_0, parameters)    #पूरे network का forward propagation करके final output Z_final निकालें
    cost = final_cost(Z_final, Y)    #Z_final और actual Y से cost/loss calculate करें
    train_net = tf.train.GradientDescentOptimizer(learning_rate).minimize(cost)    #gradient descent optimizer बनाएं जो cost को minimize करेगा (weights को update करके)
    seed = 1    #minibatch shuffling के लिए शुरुआती seed set करें
    num_minibatches = int(num_samples / mini_batch_size)    #कुल कितने minibatches बनेंगे, यह निकालें
    init = tf.global_variables_initializer()    #सभी tensorflow variables (weights, bias) को initialize करने वाला operation बनाएं
    costs = []    #हर epoch का cost store करने के लिए खाली list बनाएं
    
    with tf.Session() as sess:    #training चलाने के लिए एक tensorflow session शुरू करें
        sess.run(init)    #session में सभी variables को actually initialize करें
        for epoch in range(num_iter):    #num_iter बार (यानी इतने epochs तक) पूरे data पर training करें
            epoch_cost = 0    #इस epoch का cost शुरुआत में 0 रखें
            mini_batches = random_samples_minibatch(X_train, Y_train, mini_batch_size, seed) #random_samples_minibatch() का उपयोग करके mini-batches की array generate करें
            seed = seed + 1    #अगली बार अलग shuffling हो इसलिए seed को 1 से बढ़ाएं
            
            #perform gradient descent for each mini-batch    हर mini-batch पर gradient descent (weights update) perform करें
            for mini_batch in mini_batches:    #हर mini-batch के लिए loop चलाएं
                ##Start code
                X_batch, Y_batch = mini_batch    #इस mini_batch tuple से X_batch और Y_batch को अलग-अलग निकालें
                _,mini_batch_cost = sess.run([train_net, cost], feed_dict={A_0: X_batch, Y: Y_batch})    #placeholders में यह batch feed करके train_net (weight update) और cost दोनों को run करें
                ##End code
                
                epoch_cost += mini_batch_cost/num_minibatches    #इस batch के cost को average के हिसाब से epoch_cost में जोड़ें
            if epoch % 1 == 0:    #हर epoch पर (यानी हमेशा)
                costs.append(epoch_cost)    #इस epoch का cost costs list में save करें
            if epoch % 10 == 0:    #हर 10वें epoch पर
                print(epoch_cost)    #उस epoch का cost print करके दिखाएं

        #save your answer in pkl file. Do not modify    अपना answer pkl file में save करें, इसे modify मत करें
        with open('output/output1.pkl', 'wb') as file:    #output1.pkl नाम की file को binary write mode में खोलें
            cost_value=round(costs[-1],2)    #आखिरी epoch के cost को 2 decimal तक round करें
            print(cost_value)    #उस final cost value को print करें
            pickle.dump(cost_value, file)    #उस value को pickle से file में save करें
            
        plt.ylim(0, max(costs))    #y-axis की limit 0 से सबसे बड़े cost value तक set करें (मूल कोड में गलत तीसरा argument था, ylim सिर्फ 2 arguments लेता है इसलिए ठीक किया गया)
        plt.xlabel("epoches per 100")    #x-axis का label set करें
        plt.ylabel("cost")    #y-axis का label set करें
        plt.plot(costs)    #सभी epochs के costs की line plot बनाएं
        plt.show()    #plot को display करें
        params = sess.run(parameters)    #trained parameters (weights, bias) को actual numpy values के रूप में निकालें
    return params    #trained parameters की dictionary return करें
```

- Output predict करने वाला function आपके लिए पहले से define किया गया है  
- यह function input और trained parameters लेता है और forward propagation का उपयोग करके output predict करता है  
- arg_max function हर sample के output y में सबसे बड़ी probability value वाला index return करता है  

```python
def predict(X_train, params):    #function define करें जो input data और trained parameters लेगा
    with tf.Session() as sess:    #prediction के लिए एक नया tensorflow session शुरू करें
        Y = tf.arg_max(l_layer_forwardProp(X_train,params), dimension=0)    #forward propagation से output निकालें और हर sample के लिए सबसे बड़ी probability वाला class index (dimension=0 यानी classes के across) निकालें
        return sess.run(Y)    #session run करके predicted class indices की actual values return करें
```

Model द्वारा predict किए गए boundary को visualize करने के लिए अगले दो cells run करें (plot दिखने में थोड़ा ज्यादा समय लग सकता है)


```python
def plot_decision_boundary1( X, y, model):    #function define करें जो data X, actual labels y, और trained model (prediction function) लेगा
    plt.clf()    #पहले से बना कोई plot हो तो उसे clear (साफ) करें
    # Set min and max values and give it some padding
    x_min, x_max = X[0, :].min() - 1, X[0, :].max() + 1    #x-axis (feature1) की minimum और maximum values निकालें, थोड़ा padding (-1,+1) जोड़ें
    y_min, y_max = X[1, :].min() - 1, X[1, :].max() + 1    #y-axis (feature2) की minimum और maximum values निकालें, थोड़ा padding जोड़ें   
    colors=['blue','green','red', 'yellow']    #चारों classes के लिए colors की list
    cmap = matplotlib.colors.ListedColormap(colors)    #इन colors से custom colormap बनाएं   
    h = 0.01    #grid के points के बीच की दूरी (step size) set करें
    # Generate a grid of points with distance h between them
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))    #x और y range में h दूरी वाले points का एक meshgrid (2D grid) बनाएं
    # Predict the function value for the whole grid
    A = model(np.c_[xx.ravel(), yy.ravel()])    #grid के हर point के लिए model से prediction निकालें (xx, yy को flatten करके जोड़ा गया है)
    A = A.reshape(xx.shape)    #predictions को वापस grid जैसी shape में बदलें
    # Plot the contour and training examples
    plt.contourf(xx, yy, A, cmap="spring")    #predictions के आधार पर filled contour plot बनाएं जो decision regions दिखाएगा
    plt.ylabel('x2')    #y-axis का label set करें
    plt.xlabel('x1')    #x-axis का label set करें
    plt.scatter(X[0, :], X[1, :], c=y, s=8,cmap=cmap)    #actual data points को उनके असली labels के रंग के साथ scatter plot में दिखाएं
    plt.title("Decision Boundary")    #plot का title set करें
    plt.show()    #पूरा plot display करें
```
